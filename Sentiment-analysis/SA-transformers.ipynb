NLP (Natural Lenguage Process) for Sentiment Analysis in Spanish using transformers and torch

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
# Model selection
model_name = "edumunozsala/roberta_bne_sentiment_analysis_es"

# Loads the tokenizer and the pre-trained model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

def predict_sentiment(text):
    """Tokenizes input text and returns the predicted sentiment label."""
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=128)
    
    with torch.no_grad():
        outputs = model(**inputs)
    
    logits = outputs.logits
    prediction = torch.argmax(logits, dim=1).item()
    
    # Map the numerical prediction to a string label
    label = "NEGATIVE" if prediction == 0 else "POSITIVE"
    return label

# Execution and testing
texts = [
    "La experiencia fue increíble, volveré sin duda.",
    "No me gustó el producto, esperaba más.",
    "Está bien, pero podría ser mejor.",
    "esta bien, pero tambien esta mal"
]

for t in texts:
    print(f"Text: {t}")
    print(f"→ Sentiment: {predict_sentiment(t)}\n")
```
